```python
import math
def area_of_circle():
    pii=math.pi
    r=eval(input("enter the radius"))
    area=round(pii*r**2)
    print("the area of circle with radius {} is{}".format(r,area))
    return(area,pii)
Area,Pi=area_of_circle()
Area
```
