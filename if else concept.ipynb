```python
dis=float(input("enter distance in km"))
if dis>10: 
    print("The charge for {} km is 100rs".format(dis))
elif 7<dis<10:
    print("The charge for {} km is 70rs".format(dis))
elif 4<dis<7:
    print("The charge for {} km is 40rs".format(dis))
else:
    print("The charge is 30 rs")
```

```python
oper=input("enter which operation you need")
num1=eval(input("enter the first number"))
num2=eval(input("enter the second number"))
if oper=="add":
    print("the add of {} and {} is {}:".format(num1,num2,num1+num2))
elif oper=="sub":
    print("the sub of {} and {} is {}:".format(num1,num2,num1-num2))
elif oper=="multiplication":
    print("the multiplication of {} and {} is {}:".format(num1,num2,num1*num2))
elif oper=="divison":
    print("the divison of {} and {} is{}:".format(num1,num2,num1/num2))
else:
    print("please provide suitable values")
    
```
