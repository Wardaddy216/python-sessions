```python
#sum of 10 natural numbers
sum=0
for x in range(1,11):
    print("the sum of {} + {} is:".format(sum,x))
    sum=sum+x
    print(sum)
print("the sum of first ten natural numbers is:",sum)

    
```

```python
import random
count_odd=0
count_even=0
for i in range(5):
    random_num=random.randint(1,10)
    if random_num%2==0:
        count_even=count_even+1
        print("the number {} is even".format(random_num))
    else:
        count_odd=count_odd+1
        print("the number {} is odd".format(random_num))

print("the total number of even number is: {}".format(count_even))
print("the total number of odd number is: {}".format(count_odd))

    
```
